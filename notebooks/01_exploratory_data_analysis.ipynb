```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Load dataset
df = pd.read_csv("../data/raw/train.csv")

# Inspect dataset
print("Dataset shape:", df.shape)

df.head()
```

```python
df["SalePrice"].describe()
```

```python
print("Mean:", df["SalePrice"].mean())
print("Median:", df["SalePrice"].median())
print("Standard deviation:", df["SalePrice"].std())
print("Skewness:", df["SalePrice"].skew())
```

```python
# Data Visualization

plt.figure(figsize=(10, 5))

sns.histplot(
    df["SalePrice"],
    bins=40,
    kde=True
)

plt.title("Distribution of House Sale Prices")
plt.xlabel("Sale Price ($)")
plt.ylabel("Number of Properties")

plt.show()
```

```python
# Investiatin Missing Values

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print(missing)
```

```python
# Calculating Percentages

missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage = missing_percentage[
    missing_percentage > 0
]

print(missing_percentage)
```

```python
# Understanding Feature Types

print("Numerical columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical Columns:")
print(df.select_dtypes(include=["object"]).columns.tolist())
```

```python
df.describe(include="all")
```

```python
# Experiment A — How many rows would we lose?

# Count rows containingg atleast one missin value
rows_with_missing = df.isnull().any(axis=1).sum()

# Count rows with no missin values
complete_rows = df.dropna().shape[0]

print("Total rows:", len(df))
print("Rows with missing values:", rows_with_missing)
print("Complete rows:", complete_rows)

print(
    "Percentae of rows lost:",
    round((rows_with_missing / len(df)) * 100, 2),
    "%"
)
```

```python
# Experiment B — Identify the five most incomplete features

missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean() * 100
})

missing_summary = (
    missing_summary[missing_summary["Missing_Count"] > 0]
    .sort_values("Missing_Percentage", ascending=False)
)

missing_summary.head(10)
```

```python
# Experiment C — Investigate GarageType and PoolQC

print("GarageType:")
print(df["GarageType"].value_counts(dropna=False))

print("\nPoolQC:")
print(df["PoolQC"].value_counts(dropna=False))
```

**Feature relationships and correlation analysis.**

_What actually makes a house expensive?_

```python
# Which property characteristics have the strongest relationship with SalePrice?

# Experiment A — Find numerical correlations

# Select numerical columns
numerical_df = df.select_dtypes(include=np.number)

# Calculate Correlations
correlations = numerical_df.corr()

# Show features most correlated with SalePrice
price_correlations = (
    correlations["SalePrice"]
    .sort_values(ascending=False)
)
print(price_correlations)
```

```python
# Experiment B — Visualize the correlation matrix

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlations,
    cmap="coolwarm",
    center=0,
    annot=False
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
# Experiment C — Investigate three important features

features = ["OverallQual", "GrLivArea", "GarageCars"]

for feature in features:
    plt.figure(figsize=(7, 6))

    sns.scatterplot(
        data=df,
        x=feature,
        y="SalePrice"
    )

    plt.title(f"{feature} vs SalePrice")
    plt.show()
```

```python
# Experiment D — Investigate outliers

plt.figure(figsize=(10, 5))

sns.boxplot(x = df["SalePrice"])

plt.title("SalePrice Outlier Analysis")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="GrLivArea",
    y="SalePrice"
)

plt.title("Living Area vs Sale Price")
plt.show()
```

```python
# Difference btn Pearson and Spearman Correlation. .corr() uses Pearson

# Pearson correlation
pearson_corr = df.select_dtypes(include=np.number).corr(method="pearson")

# Spearman correlation
spearman_corr = df.select_dtypes(include=np.number).corr(method="spearman")

print("Pearson:")
print(pearson_corr["SalePrice"].sort_values(ascending=False).head(10))

print("\nSpearman:")
print(spearman_corr["SalePrice"].sort_values(ascending=False).head(10))
```

```python
# Inspect properties with the largest above-round livin ares

largest_houses = df.nlargest(10, "GrLivArea")[
    ["Id", "GrLivArea", "OverallQual", "GarageCars", "SalePrice"]
]

largest_houses
```
